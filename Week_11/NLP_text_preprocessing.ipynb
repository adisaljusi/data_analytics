## Text preprocessing
#### Steps:
- Text to lowercase
- Removing punctuations
- Tokenization
- Removal of stop words
- Lemmatization

### Text to lowercase

```python
# Text to lowercase function
def text_lowercase(text):
    return text.lower()

# Text to lowercase
corpus_02 = text_lowercase(corpus_01)
corpus_02
```

### Removing punctuation

```python
# Remove punctuation function
def remove_punctuation(text):
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)

# Remove punctuation
corpus_03 = remove_punctuation(corpus_02)
corpus_03
```

### Tokenize text & removal of stopwords

```python
# Show english stopwords
eng_stopwords = set(stopwords.words('english'))
print("List of english stopwords:")
print(eng_stopwords)
```

```python
# Function for tokenization and the removal of stopwords
def remove_stopwords(text):
    stop_words = set(stopwords.words("english"))
    word_tokens = word_tokenize(text)
    filtered_text = [word for word in word_tokens if word not in stop_words]
    return filtered_text
 
# Remove stopwords
corpus_04 = remove_stopwords(corpus_03)
print(corpus_04, end="")
```

### Lemmatization

```python
# Initialize Lemmatizer
lemmatizer = WordNetLemmatizer()

# Lemmatize string function
def lemmatize_word(text):
    word_tokens = word_tokenize(text)
    lemmas = [lemmatizer.lemmatize(word, pos ='v') for word in word_tokens]
    return lemmas

# Lemmatize
lem = []
for i in corpus_04:
    lem.append(lemmatize_word(i))

# Nested list to list
corpus_05 = [' '.join([str(x) for x in lst]) for lst in lem]

print('Before lemmatization:')
print(corpus_04, '\n')

print('After lemmatization:')
print(corpus_05, end="")
```

## Term frequency-inverse document frequency (TF-IDF)
- For details see: https://www.learndatasci.com/glossary/tf-idf-term-frequency-inverse-document-frequency

### Term Frequency (TF)

```python
# Compute Term Frequency (TF)
words_set = set()
for doc in corpus:
    words = doc.split(' ')
    words_set = words_set.union(set(words))
    
print('Number of words in the corpus:',len(words_set), '\n')
print('The words in the corpus: \n', words_set)

# Number of documents in the corpus
n_docs = len(corpus)

# Number of unique words in the corpus 
n_words_set = len(words_set)

df_tf = pd.DataFrame(np.zeros((n_docs, n_words_set)), 
                     columns=list(words_set))

print("\nTerm Frequency (TF):")
for i in range(n_docs):
    # Words in the document
    words = corpus[i].split(' ')
    for w in words:
        df_tf[w][i] = df_tf[w][i] + (1 / len(words))
        
print(df_tf.round(4))
```

### Inverse Document Frequency (IDF)

```python
# Computing Inverse Document Frequency (IDF)
print("\nInverse Document Frequency (IDF):")

idf = {}

for w in words_set:
    
    # k = number of documents that contain this word
    k = 0
    
    for i in range(n_docs):
        if w in corpus[i].split():
            k += 1
            
    idf[w] =  np.log10(n_docs / k).round(4)
    
    print(f'{w:>15}: {idf[w]:>10}')
```

### Term Frequency - Inverse Document Frequency (TF-IDF)

```python
# Computing TF-IDF
df_tf_idf = df_tf.copy()

for w in words_set:
    for i in range(n_docs):
        df_tf_idf[w][i] = df_tf[w][i] * idf[w]

print('\nTF-IDF:')
print(df_tf_idf.round(4))
```

## Task 1e: TF, IDF, and TF-IDF matrices for custom corpus

```python
# Task 1e: Compute TF for custom corpus
words_set_custom = set()
for doc in corpus_custom:
    words = doc.split(' ')
    words_set_custom = words_set_custom.union(set(words))
    
print('Number of words in custom corpus:', len(words_set_custom), '\n')
print('Words in custom corpus:', words_set_custom, '\n')

# Number of documents and unique words
n_docs_custom = len(corpus_custom)
n_words_set_custom = len(words_set_custom)

# Create TF dataframe
df_tf_custom = pd.DataFrame(np.zeros((n_docs_custom, n_words_set_custom)), 
                           columns=list(words_set_custom))

print("Term Frequency (TF) for custom corpus:")
for i in range(n_docs_custom):
    words = corpus_custom[i].split(' ')
    for w in words:
        df_tf_custom[w][i] = df_tf_custom[w][i] + (1 / len(words))
        
print(df_tf_custom.round(4))
```

```python
# Task 1e: Compute IDF for custom corpus
print("\nInverse Document Frequency (IDF) for custom corpus:")

idf_custom = {}

for w in words_set_custom:
    k = 0
    for i in range(n_docs_custom):
        if w in corpus_custom[i].split():
            k += 1
    idf_custom[w] = np.log10(n_docs_custom / k).round(4)
    print(f'{w:>15}: {idf_custom[w]:>10}')
```

```python
# Task 1e: Compute TF-IDF for custom corpus
df_tf_idf_custom = df_tf_custom.copy()

for w in words_set_custom:
    for i in range(n_docs_custom):
        df_tf_idf_custom[w][i] = df_tf_custom[w][i] * idf_custom[w]

print('\nTF-IDF for custom corpus:')
print(df_tf_idf_custom.round(4))
```

## Part-of-Speach (POS) tagging
For meaning of POS-tags see: https://pythonexamples.org/nltk-pos-tagging

```python
text = '''European authorities fined Google a record $5.1 
          billion on Wednesday for abusing its power in the 
          mobile phone market.'''

def preprocess(sent):
    sent = nltk.word_tokenize(sent)
    sent = nltk.pos_tag(sent)
    return sent

sent = preprocess(text)
pattern = 'NP: {<DT>?<JJ>*<NN>}'

cp = nltk.RegexpParser(pattern)
cs = cp.parse(sent)

iob_tagged = tree2conlltags(cs)

# Print the POS-tags
pprint(iob_tagged)
```

```python
# Task 1f: POS tagging with custom text
text_custom = '''Artificial intelligence and machine learning are revolutionizing 
                 various industries including healthcare, finance, and transportation. 
                 These technologies enable businesses to automate processes and make 
                 data-driven decisions.'''

# Preprocess the custom text
sent_custom = preprocess(text_custom)

# Define pattern
pattern_custom = 'NP: {<DT>?<JJ>*<NN>}'

# Parse
cp_custom = nltk.RegexpParser(pattern_custom)
cs_custom = cp_custom.parse(sent_custom)

# Get IOB tags
iob_tagged_custom = tree2conlltags(cs_custom)

# Print the POS-tags
print('POS tags for custom text:')
pprint(iob_tagged_custom)
```

### Explanation of POS Tags (Task 1f)

Based on the [NLTK POS tagging reference](https://pythonexamples.org/nltk-pos-tagging), here are explanations for common POS tags in the output:

1. **NN (Noun, singular)**: Refers to a singular noun. Example: "intelligence", "learning", "healthcare"
   - These are common nouns that represent a single entity or concept

2. **NNS (Noun, plural)**: Refers to plural nouns. Example: "industries", "technologies", "businesses"
   - These represent multiple instances of a noun

3. **JJ (Adjective)**: Describes or modifies a noun. Example: "Artificial", "various", "data-driven"
   - Adjectives provide additional information about the characteristics of nouns

4. **VBG (Verb, gerund/present participle)**: A verb form ending in "-ing" that can function as a noun or adjective. Example: "revolutionizing", "including"
   - Often used to describe ongoing actions or processes

5. **VBP (Verb, non-3rd person singular present)**: A present tense verb form used with I/you/we/they. Example: "are", "enable", "make"
   - These are action words in present tense

6. **CC (Coordinating conjunction)**: Connects words, phrases, or clauses. Example: "and"
   - Used to join similar elements in a sentence

7. **IN (Preposition or subordinating conjunction)**: Shows relationships between nouns/pronouns and other words. Example: "to"
   - Indicates location, direction, time, or other relationships

8. **DT (Determiner)**: Introduces a noun phrase. Example: "These"
   - Articles and demonstratives that specify which noun is being referenced
