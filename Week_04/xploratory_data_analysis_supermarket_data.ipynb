```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('supermarkets_data_enriched.csv')

df.head()
```

# Count: Supermarkets per brand

```python
brand_counts = df['brand'].value_counts()
brand_counts
```

# Bar chart of supermarkets per brand

```python
plt.figure(figsize=(18,10))
brand_counts.plot(kind='bar', color='blue')
plt.title('Number of supermarkets per brand', fontsize=12)
plt.xlabel('Brand')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

# Pairgrid - Scatterplot matrix

```python
sns.set_theme(style="ticks", font_scale=0.8)
cols = ['lat', 'lon', 'pop', 'pop_dens', 'frg_pct', 'emp']

g = sns.PairGrid(df[cols].dropna())
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot, s=20)
plt.show()
```

# Plotting the locations on a map

```python
subset = df[df['brand'].isin(['Denner', 'Volg', 'Landi'])].copy()

# Create the plot with improved styling
plt.figure(figsize=(10, 8))
sns.scatterplot(data=subset, x='lon', y='lat', hue='brand', s=50,  palette='deep')

plt.title('Supermarket Locations: Denner, Volg, and Landi', fontsize=14, fontweight='bold')
plt.xlabel('Longitude', fontsize=11)
plt.ylabel('Latitude', fontsize=11)
plt.legend(title='Brand', title_fontsize=11, fontsize=10)
plt.grid(True, alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```
