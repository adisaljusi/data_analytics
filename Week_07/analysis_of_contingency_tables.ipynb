# Analysis of contingency tables

## Libraries and settings

```python
# Libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2
from scipy.stats import chi2_contingency

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
```

## Import the apartment data

```python
# Define columns
columns = ['web-scraper-order',
            'address_raw',
            'lat',
            'lon',
            'bfs_number',
            'bfs_name',
            'rooms', 
            'area', 
            'luxurious', 
            'price', 
            'price_per_m2',
            'pop_dens',
            'frg_pct',
            'mean_taxable_income',
            'dist_supermarket']

# Read and select variables
df_orig = pd.read_csv("apartments_data_enriched_cleaned.csv", sep=";")[columns]

# Rename variable 'web-scraper-order' to 'apmt_id'
df_orig = df_orig.rename(columns={'web-scraper-order': 'apartment_id'})

# Remove missing values
df = df_orig.dropna()
df.head(5)

# Remove duplicates
df = df.drop_duplicates()
df.head(5)
```

### Create new categorical variable 'area_cat'

```python
# Create labels
labels = ['0 - 49', '50 - 74', '75 - 99', '100 - 149', '>= 150']

# Create new categorical variable
df["area_cat"] = pd.cut(df['area'], bins=[0, 50, 75, 100, 150, 1000], labels=labels)

# Check values
df[['area', 'area_cat']].head(10)
```

## Create a contingency table with the number of apartments per 'area_cat' versus 'price_cat'

```python
# Contingency table (numbers in the table are apartments)
tab = pd.pivot_table(df[['apartment_id', 'area_cat', 'price_cat']],
                     index=['area_cat'],
                     columns=['price_cat'], 
                     aggfunc='count')
tab
```

## Perform a Chi-square test on the contingency table
For details see: https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.chi2_contingency.html

```python
# 1-alpha
prob = 0.95

# Chi-squared test
stat, p, dof, expected = chi2_contingency(tab)

# Critical value
critical = chi2.ppf(prob, dof)

# Print results of the Chi-square test
print('Expected frequencies:', '\n', expected.round(4), '\n')
print('Critical value:', critical.round(4), '\n')
print('Test statistic:', stat.round(4), '\n')
print('Degrees of freedom:', dof, '\n')
print(f'p-value: {p:.4f}')
```

## Interpretation of the Chi-squared test result

<p>Because the test statistic is larger than the critical value and the p-value is smaller than our significance level apha=0.05, we can reject the null hypothesis, i.e. there is a statistically significant (at the 0.05 significance-level) association between the price-categories and the living-area categories in the contingence table.</p>

### Contingency table, spotted hyena example

```python
# The numpy array shows how many times each hyena responded (see study above)
table = np.array([[0,4],
                  [7,6],
                  [9,0],
                  [15,10]])

# Create a data frame from 'table'
df_hy = pd.DataFrame(table, columns=[['Female', 'Male']],
             index=['Fear', 'Greeting', 'Aggression', 'No Response'])
df_hy
```

### Perform Chi-squared test on hyena behavior data

```python
prob = 0.95

stat_hy, p_hy, dof_hy, expected_hy = chi2_contingency(table)

critical_hy = chi2.ppf(prob, dof_hy)

print('Expected frequencies:', '\n', expected_hy.round(4), '\n')
print('Critical value:', critical_hy.round(4), '\n')
print('Test statistic:', stat_hy.round(4), '\n')
print('Degrees of freedom:', dof_hy, '\n')
print(f'p-value: {p_hy:.4f}')
```

### Python function to calculate expected frequencies

```python
def calculate_expected_frequencies(observed):
    if isinstance(observed, pd.DataFrame):
        observed = observed.values
    
    row_totals = observed.sum(axis=1)
    col_totals = observed.sum(axis=0)
    grand_total = observed.sum()
    
    expected = np.outer(row_totals, col_totals) / grand_total
    
    return expected

expected_manual = calculate_expected_frequencies(table)
print('Expected frequencies (manual calculation):')
print(expected_manual.round(4))
```

### Compare manual calculation with chi2_contingency() expected frequencies

```python
# Compare manual calculation with chi2_contingency()
print('Expected frequencies from chi2_contingency():')
print(expected_hy.round(4))
print('\n')
print('Expected frequencies from manual calculation:')
print(expected_manual.round(4))
print('\n')
print('Difference between the two methods:')
print((expected_hy - expected_manual).round(8))
```

### Extended function to calculate Chi-square test statistic

```python
def calculate_chi_square_statistic(observed):
    """
    Calculate expected frequencies and Chi-square test statistic for a contingency table.
    
    Parameters:
    -----------
    observed : numpy array or pandas DataFrame
        The observed frequencies in the contingency table
    
    Returns:
    --------
    tuple: (expected_frequencies, chi_square_statistic)
        The expected frequencies and the Chi-square test statistic
    """
    # Convert to numpy array if it's a DataFrame
    if isinstance(observed, pd.DataFrame):
        observed = observed.values
    
    # Calculate row totals and column totals
    row_totals = observed.sum(axis=1)
    col_totals = observed.sum(axis=0)
    grand_total = observed.sum()
    
    # Calculate expected frequencies
    expected = np.outer(row_totals, col_totals) / grand_total
    
    # Calculate Chi-square statistic: χ² = Σ[(O - E)² / E]
    chi_square = np.sum((observed - expected)**2 / expected)
    
    return expected, chi_square

# Apply the function to hyena data
expected_calc, chi_square_calc = calculate_chi_square_statistic(table)
print('Expected frequencies (manual calculation):')
print(expected_calc.round(4))
print('\n')
print(f'Chi-square test statistic (manual calculation): {chi_square_calc:.4f}')
```

### Compare manual Chi-square statistic with chi2_contingency() result

```python
# Compare manual calculation with chi2_contingency()
print(f'Chi-square statistic from chi2_contingency(): {stat_hy:.4f}')
print(f'Chi-square statistic from manual calculation: {chi_square_calc:.4f}')
print('\n')
print(f'Difference: {abs(stat_hy - chi_square_calc):.8f}')
```

### Interpretation: Are the results of the original study correct?

Based on the Chi-squared test results (p-value < 0.05), the original study correctly concluded that there is a statistically significant association between the type of behavior and the gender of the spotted hyenas, indicating that female and male hyenas respond differently to social interactions.
